```python
import pandas as pd
import numpy as nn
import matplotlib.pyplot as plt
# 1.加载数据
data1 = pd.read_excel('meal_order_detail.xlsx', sheet_name = 'meal_order_detail1')
data2 = pd.read_excel('meal_order_detail.xlsx', sheet_name = 'meal_order_detail1')
data3 = pd.read_excel('meal_order_detail.xlsx', sheet_name = 'meal_order_detail1')
```

```python
#数据预处理（合并数据）
data = pd.concat([data1,data2,data3], axis = 0)
data.head(5)
```

```python
data.info()
# 删除控制列
data.dropna(axis = 1, inplace = True)
```

```python
data.info()
```

```python
#卖出菜品的平均价格， 平均可以代表一类菜品，也可以作为一种全局标准。
#numpy里面的平均值计算寒素比pandas自带函数计算效率更快。
round(data['amounts'].mean(),2)
round(nn.mean(data['amounts']),2) #操作更快，更适合大数据。
```

```python
#什么菜最受欢迎。
dishes_count = data['dishes_name'].value_counts()[:10]
dishes_count
```

```python
#可视化（一维）
dishes_count.plot(kind = 'bar')
```
