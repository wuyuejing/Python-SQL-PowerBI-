```python
import pandas as pd
```

```python
df = pd.read_excel('./meal_order_detail.xlsx')
df
```

```python
def 菜分类(x):
    if '菜' in x:
        type_str = '菜'
    elif '汤' in x:
        type_str = '汤'
    else:
        type_str = '其他'
    return type_str
```

```python
(
    df
    .assign(菜类型=lambda d:d.dishes_name.map(菜分类))
    .query("菜类型 in ['菜', '汤']")
    .loc[:,["place_order_time", "菜类型", "amounts"]]
    .pivot_table(index = "place_order_time", columns = "菜类型", values = "amounts", aggfunc='sum' )
    .sum()
    #.shape
)
```
