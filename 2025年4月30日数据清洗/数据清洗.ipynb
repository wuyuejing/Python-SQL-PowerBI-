```python
#1. 缺失值处理
import pandas as pd
df = pd.read_excel('数据预处理.xlsx', sheets_name="考核表")
df
```

```python
#删空
#存在一个缺失值一行删除
df.dropna()
#整行为缺失值才删除
df.dropna(how = "all")
```

```python
#填充
#填充统一一个值
df.fillna(0)
df.fillna('未知')
#根据每列特性自定义填充方式
df.fillna(pd.Series(['无名氏', 0, '待定'], ['姓名', '分数', '状态']))#前为值，后为列标题
df.fillna({'姓名':'无名氏', '分数'：0, '状态':'待定'})#字典格式
```

```python
#2.异常值替换
import pandas as pd
df = pd.read_excel('数据预处理.xlsx', sheets_name="学生成绩表")
df
```

```python
#单值替换
df.replace(0, 10)#值
df.replace('待定', '有数')#字符
#多值替换
df.replace(['小仙','小限'], '小贤')#单列多值
df.replace([118, '麻子'], [102, '小贤']) #多列多值
df.replace({108: 118, '小闲': '小贤'})#字典格式
#改原表：inplace=true
```
